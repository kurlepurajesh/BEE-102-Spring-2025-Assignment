## Markov Transition Matrix Construction from DNA Sequence


### Approach:

1. **Define States:**
   - The four valid nucleotide states considered are: A, C, G, and T.

2. **Input Validation:**
   - Before constructing the matrix, the sequence is validated to ensure that it contains only valid characters (A, C, G, T).
   - If any invalid nucleotide is found, an error is raised to prevent incorrect results.

3. **Counting Transitions:**
   - For every pair of adjacent nucleotides in the sequence (i.e., dinucleotides), a count is maintained of how many times a transition occurs from one nucleotide to another.
   - These counts are stored in a nested dictionary: `counts[current][next] += 1`.

4. **Normalization:**
   - The transition counts are then converted into probabilities by dividing each count by the total number of transitions from that nucleotide.
   - This step results in a proper **Markov transition matrix**, where each row sums to 1 (or 0 if no transitions exist for that nucleotide).

5. **Conversion to DataFrame:**
   - The final matrix is stored and displayed as a **pandas DataFrame** for clear tabular visualization.
   - Rows represent the current nucleotide, and columns represent the next nucleotide.

### Why Build a Markov Model?

- A **simple dinucleotide frequency model** only counts how often a specific pair of nucleotides appears, without considering the directionality or context of transitions.
- In contrast, a **Markov model** captures the **conditional probability** of moving from one state (nucleotide) to the next.
- This allows us to model **dependencies** in sequences and is particularly useful in biological sequence analysis where context matters (e.g., promoter prediction, gene structure modeling).
- The transition matrix built using a Markov model can be used to simulate DNA sequences or compute likelihoods of observed sequences under specific probabilistic assumptions.

### Output:

- The printed **Markov transition matrix** displays how likely each nucleotide is to follow another.
- This can be directly interpreted to understand sequence biases and patterns.

```python
import pandas as pd

def build_markov_transition_matrix(sequence):
    """
    Build the Markov transition matrix for a given DNA sequence.

    Args:
    sequence (str): Input DNA sequence composed of the nucleotides A, C, G, T.

    Returns:
    pd.DataFrame: Markov transition matrix.
    """

    # Define the nucleotide states (A, C, G, T)
    states = ['A', 'C', 'G', 'T']

    # Initialize the dictionary for counts of transitions
    counts = {nucleotide: {next_nucleotide: 0 for next_nucleotide in states} for nucleotide in states}

    # Check if sequence contains only valid nucleotides (A, C, G, T)
    if not all(nuc in states for nuc in sequence):
        raise ValueError("Invalid nucleotide encountered in sequence. Only A, C, G, T are allowed.")

    # Count occurrences of each dinucleotide (transition between nucleotides)
    for i in range(len(sequence) - 1):
        current_nucleotide = sequence[i]
        next_nucleotide = sequence[i + 1]
        if current_nucleotide in states and next_nucleotide in states:
            counts[current_nucleotide][next_nucleotide] += 1

    # Convert counts to probabilities by normalizing each row
    transition_matrix = {nucleotide1: {} for nucleotide1 in states}

    for nucleotide1 in states:
        total_transitions = sum(counts[nucleotide1].values())

        if total_transitions > 0:
            for nucleotide2 in states:
                transition_matrix[nucleotide1][nucleotide2] = counts[nucleotide1][nucleotide2] / total_transitions
        else:
            # If no transitions from a given nucleotide, assign zero probability for each
            for nucleotide2 in states:
                transition_matrix[nucleotide1][nucleotide2] = 0.0

    # Convert the dictionary to a pandas DataFrame for easy visualization
    markov_matrix = pd.DataFrame(transition_matrix)

    return markov_matrix

# Example DNA sequence (you can change this to any valid sequence)
sequence = (
    "ATGCGTACGTTAGCTAGCATCGATCGGATCGATCGTACGATCGATGCTAGCTAGCTAGC"
    "TAGCTAGCATCGATCGTACGTAGCATCGATCGTAGCTAGCTAGCGTACGTAGCTAGC"
)
# Build the Markov transition matrix
try:
    markov_matrix = build_markov_transition_matrix(sequence)
    print("Markov Transition Matrix: \nrows: current base\ncolumns: next base")
    print(markov_matrix)
except ValueError as e:
    print(e)
```
