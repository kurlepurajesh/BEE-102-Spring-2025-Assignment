# **Problem Statement**

---



**Multi-line sequence fasta to single-line fasta file**

Fasta file format is shown below. Many times, it is annoying to find sequences of interest because of fix-width format, i.e., if 60 character fix-width file is there then only except the header (starting with >) of the sequence each line will have 60 characters. So, if a sequence is of width 600, then it would be written in 10 lines. Your goal is convert this format to a format with header in a single line followed by its sequence in a single line.




## **Multi-line FASTA to Single-line FASTA Format**



### Approach

1. **Read the Input File:**
   - Open the original `multiline_input.fasta` file line-by-line.

2. **Parsing Logic:**
   - If a line starts with `>`, it's a header line. Store it.
     - If there's a previously stored header and sequence, write them to the output file.
   - If the line is not a header, treat it as part of the sequence and concatenate it (after stripping newline characters).
   - After reaching the end of the file, write the last stored header and sequence.

3. **Write to Output File:**
   - Save the transformed content into a new file named `output.fasta`.

---

### Benefits

- Each record (header + sequence) now occupies exactly two lines.
- Simplifies pattern searching within sequences.
- Easier to process using string operations or bioinformatics tools.

---

### Output File

The transformed sequences are saved in:

output.fasta

Each sequence now appears in a single line, directly following its header.

---

### Summary

This transformation is useful for cleaning FASTA files for downstream tasks, especially where full sequence integrity and easier readability/search are required.

```python
input_file = "multiline_input.fasta"
output_file = "output.fasta"

with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
    header = None
    sequence = ""

    for line in infile:
        line = line.strip()
        if not line:
            continue  # Skip blank lines
        if line.startswith(">"):
            if header:
                outfile.write(header + '\n')
                outfile.write(sequence + '\n')
            header = line
            sequence = ""
        else:
            sequence += line
    # Write the last sequence
    if header:
        outfile.write(header + '\n')
        outfile.write(sequence + '\n')

print("Multi-line FASTA has been successfully converted to single-line format.")
print(f"📁 Output has been saved to: '{output_file}'")
print("\n📄 Output content (Single-line FASTA format):\n")

with open(output_file, 'r') as f:
    print(f.read())
```
